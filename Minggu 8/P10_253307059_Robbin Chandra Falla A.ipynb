  2. **Terapkan Model yang Bisa Dijelaskan (*Explainable AI*):** Hindari model *black box* tanpa transparansi. Gunakan teknik interpretasi seperti *SHAP* atau *Feature Importance* agar sistem bisa memaparkan alasan logis penolakan (misal: "Rasio beban tagihan bulanan terhadap estimasi saldo terlalu tinggi").
  3. **Penerapan *Human-in-the-Loop* & Mekanisme Banding:** Keputusan krusial yang berdampak finansial tidak boleh 100% otomatis. Sediakan jalur sanggahan/banding ke analis kredit manusia agar nasabah yang dirugikan bisa memberikan dokumen bukti keuangan pendukung.

### 4. Jelaskan mengapa “model dengan akurasi tinggi” belum tentu “sistem yang baik” — kaitkan dengan minimal tiga bab berbeda dalam buku ini.

Akurasi hanyalah metrik angka di atas kertas data uji statis, bukan jaminan sistem bekerja aman dan bermanfaat di dunia nyata. Hal ini terlihat dari konsep di 3 bab buku ini:

1. **Kaitan Bab 7 (Machine Learning — Jebakan Data Timpang / Class Imbalance):**  
   Pada data yang sangat timpang (contoh: deteksi transaksi penipuan bank di mana 99% transaksi normal dan 1% penipuan), model bodoh yang selalu menebak "Normal" akan mencatat akurasi 99%. Namun sistem ini tidak berguna sama sekali karena meloloskan 100% aksi pembobolan rekening (*Recall = 0%*).
2. **Kaitan Bab 4 & 10 (Sistem Pakar & Isu Akuntabilitas — Bahaya Model Black Box):**  
   Model deep learning modern bisa saja mencapai akurasi 96%, tetapi jika pengembang dan dokter tidak bisa menjelaskan alasan di balik rekomendasinya (*black box*), model tersebut berbahaya digunakan pada kasus medis kritis atau hukum. Sebagaimana prinsip sistem pakar (Bab 4), transparansi penalaran `IF-THEN` sering kali jauh lebih bernilai dibanding akurasi tinggi yang buta konteks.
3. **Kaitan Bab 10 (Deployment & MLOps — Masalah Model Drift di Lapangan):**  
   Akurasi tinggi saat diuji di notebook hanyalah cerminan data masa lalu. Ketika sistem dideploy ke server produksi, pola perilaku masyarakat bergeser (*model drift*). Tanpa pemantauan MLOps dan perawatan *technical debt*, model yang dulunya berakurasi 95% bisa anjlok performanya dan merugikan operasional sistem.

### Kegiatan 1 — Chatbot FAQ Tanpa LLM (±60 menit)

```python
faq = [
    {
        "kunci": ["krs", "isi", "kapan"],
        "jawab": "Pengisian KRS dibuka minggu pertama semester melalui SIAKAD.",
    },
    {
        "kunci": ["cuti", "kuliah", "syarat"],
        "jawab": "Ajukan surat cuti ke BAAK dengan persetujuan dosen wali.",
    },
    {
        "kunci": ["beasiswa", "daftar"],
        "jawab": "Info beasiswa diumumkan di web kemahasiswaan tiap awal semester.",
    },
    {
        "kunci": ["wisuda", "syarat", "daftar"],
        "jawab": "Syarat wisuda: lulus semua MK, bebas pustaka, dan bebas administrasi.",
    },
]

def chatbot_kunci(pertanyaan):
    kata = pertanyaan.lower().split()

    skor_terbaik = 0
    jawaban = "Maaf, saya belum tahu jawabannya."

    for item in faq:
        skor = sum(1 for k in item["kunci"] if k in kata)
        if skor > skor_terbaik:
            skor_terbaik = skor
            jawaban = item["jawab"]

    return jawaban

print(chatbot_kunci("Kapan saya bisa isi KRS?"))
print(chatbot_kunci("Bagaimana cara mengambil rencana studi?"))
```

### Pengembangan 1

```python
kamus_sinonim = {
    "rencana studi": "krs",
    "kartu rencana studi": "krs",
    "ambil": "isi",
    "mengambil": "isi",
    "jadwal": "kapan",
    "libur": "cuti",
    "berhenti sementara": "cuti",
    "bantuan dana": "beasiswa",
    "kelulusan": "wisuda",
}

def chatbot_kunci_plus(pertanyaan):
    # 1. Normalisasi teks: huruf kecil & bersihkan tanda baca
    kalimat = pertanyaan.lower()
    for tanda in ["?", "!", ".", ",", ":", ";"]:
        kalimat = kalimat.replace(tanda, " ")
        
    # 2. Ganti frasa sinonim dengan kata kunci standar FAQ
    for frasa, pengganti in kamus_sinonim.items():
        if frasa in kalimat:
            kalimat = kalimat.replace(frasa, pengganti)
            
    kata = kalimat.split()
    
    skor_terbaik = 0
    jawaban = "Maaf, saya belum tahu jawabannya. Silakan hubungi BAAK."
    kunci_cocok = []

    # 3. Pencocokan kata kunci
    for item in faq:
        cocok = [k for k in item["kunci"] if k in kata]
        if len(cocok) > skor_terbaik:
            skor_terbaik = len(cocok)
            jawaban = item["jawab"]
            kunci_cocok = cocok

    return jawaban, skor_terbaik, kunci_cocok

# Pengujian: Pertanyaan yang sebelumnya GAGAL
uji_tanya = [
    "Kapan saya bisa isi KRS?",
    "Bagaimana cara mengambil rencana studi?",  # Sebelumnya gagal karena beda kata
    "Apa saja syarat untuk pengajuan kelulusan?",
]

print("=== UJI CHATBOT FAQ KATA KUNCI (DENGAN SINONIM & NORMALISASI) ===\n")
for q in uji_tanya:
    jawab, skor, cocok = chatbot_kunci_plus(q)
    print(f"Tanya : {q}")
    print(f"Jawab : {jawab}")
    print(f"Info  : Skor kecocokan = {skor} (Kata cocok: {cocok})\n")
```

### Kegiatan 3 — Memanggil LLM dari Python (±45 menit, perlu API key gratis)

```python
import os
from pathlib import Path

# 1. Ambil API key otomatis dari file .env
try:
    from dotenv import load_dotenv
    load_dotenv()
except ImportError:
    pass

API_KEY = os.getenv("GEMINI_API_KEY", "")

# Fallback: jika belum terbaca di environment, cari langsung di file .env
if not API_KEY:
    for env_path in [Path(".env"), Path("../.env")]:
        if env_path.exists():
            with open(env_path, "r", encoding="utf-8") as f:
                for line in f:
                    if "GEMINI_API_KEY" in line and "=" in line:
                        API_KEY = line.split("=", 1)[1].strip().strip("'\"")
                        os.environ["GEMINI_API_KEY"] = API_KEY
                        break
            if API_KEY:
                break

# 2. Panggil LLM Gemini
if not API_KEY:
    print("Sel API dilewati -- pastikan GEMINI_API_KEY sudah terisi di file .env.")
else:
    try:
        try:
            from google import genai
            client = genai.Client(api_key=API_KEY)
            
            # Coba model yang aktif
            balasan = None
            for m in ["gemini-2.5-flash", "gemini-1.5-flash", "gemini-3.5-flash"]:
                try:
                    balasan = client.models.generate_content(
                        model=m,
                        contents="Jelaskan fuzzy logic dalam 2 kalimat"
                    )
                    break
                except Exception:
                    continue
            
            if balasan:
                print("Koneksi API Berhasil! Jawaban Gemini:")
                print(balasan.text)
            else:
                print("Panggilan API gagal: Kuota model sedang sibuk atau tidak tersedia.")
                
        except ImportError:
            # Prioritas 2: Fallback ke SDK Lama (google-generativeai)
            import google.generativeai as legacy_genai
            legacy_genai.configure(api_key=API_KEY)
            model = legacy_genai.GenerativeModel("gemini-1.5-flash")
            balasan = model.generate_content("Jelaskan fuzzy logic dalam 2 kalimat")
            print("Koneksi API Berhasil! Jawaban Gemini:")
            print(balasan.text)

    except Exception as e:
        print(f"Panggilan API gagal: {e}")
```

### Pengembangan 3 

```python
def panggil_gemini_kustom(prompt, temperature=0.2, max_tokens=120):
    """
    Memanggil LLM dengan parameter temperatur untuk mengontrol tingkat kebebasan/kreativitas respon.
    Dukungan otomatis untuk google-genai (baru) dan google-generativeai (lama).
    """
    if not API_KEY:
        return "[Simulasi] API Key belum diset di .env."
    
    try:
        try:
            from google import genai
            from google.genai import types
            client = genai.Client(api_key=API_KEY)
            
            cfg = types.GenerateContentConfig(
                temperature=temperature,
                max_output_tokens=max_tokens
            )
            for m in ["gemini-2.5-flash", "gemini-1.5-flash", "gemini-3.5-flash"]:
                try:
                    res = client.models.generate_content(
                        model=m,
                        contents=prompt,
                        config=cfg
                    )
                    return res.text.strip()
                except Exception:
                    continue
            return "[Gagal menemukan model yang aktif]"
            
        except ImportError:
            import google.generativeai as legacy_genai
            legacy_genai.configure(api_key=API_KEY)
            cfg = legacy_genai.types.GenerationConfig(
                temperature=temperature,
                max_output_tokens=max_tokens
            )
            model = legacy_genai.GenerativeModel("gemini-3.5-flash", generation_config=cfg)
            res = model.generate_content(prompt)
            return res.text.strip()
            
    except Exception as e:
        return f"[Gagal: {e}]"

# Pengujian perbandingan temperatur dengan prompt yang sama
prompt_uji = "Jelaskan konsep neural network dalam 1 analogi sederhana."

print("=== PENGUJIAN KONTROL TEMPERATURE GEMINI ===")
print("\n--- 1. Temperature = 0.1 (Faktual & Deterministik) ---")
print(panggil_gemini_kustom(prompt_uji, temperature=0.1))

print("\n--- 2. Temperature = 0.9 (Kreatif & Beragam Gaya Bahasa) ---")
print(panggil_gemini_kustom(prompt_uji, temperature=0.9))
```

```python
# 1. Basis Data FAQ Lengkap: 4 FAQ Awal + 3 FAQ Baru yang Ditambahkan
faq_mandiri = [
    # FAQ Awal Kegiatan 1
    {
        "kunci": ["krs", "isi", "kapan"],
        "jawab": "Pengisian KRS dibuka minggu pertama semester melalui SIAKAD.",
    },
    {
        "kunci": ["cuti", "kuliah", "syarat"],
        "jawab": "Ajukan surat cuti ke BAAK dengan persetujuan dosen wali.",
    },
    {
        "kunci": ["beasiswa", "daftar"],
        "jawab": "Info beasiswa diumumkan di web kemahasiswaan tiap awal semester.",
    },
    {
        "kunci": ["wisuda", "syarat", "daftar"],
        "jawab": "Syarat wisuda: lulus semua MK, bebas pustaka, dan bebas administrasi.",
    },
    # 3 FAQ Baru
    {
        "kunci": ["pustaka", "perpus", "buku", "pinjam"],
        "jawab": "Peminjaman buku perpustakaan maksimal 3 buku selama 1 minggu, perpanjangan mandiri via web perpustakaan.",
    },
    {
        "kunci": ["magang", "pkl", "industri", "kerja", "daftar"],
        "jawab": "Pendaftaran magang industri / PKL dibuka pada semester 5 melalui koordinator magang program studi.",
    },
    {
        "kunci": ["legalisir", "ijazah", "transkrip", "dokumen"],
        "jawab": "Permohonan legalisir ijazah dan transkrip nilai dapat diajukan secara online melalui portal layanan BAAK.",
    },
]

# 2. Kamus Pemetaan Sinonim untuk Menangani Variasi Kata Mahasiswa
kamus_sinonim_mandiri = {
    "rencana studi": "krs",
    "kartu rencana studi": "krs",
    "ambil": "isi",
    "mengambil": "isi",
    "jadwal": "kapan",
    "libur": "cuti",
    "berhenti sementara": "cuti",
    "bantuan dana": "beasiswa",
    "dana bantuan": "beasiswa",
    "kelulusan": "wisuda",
    "toga": "wisuda",
    "perpustakaan": "perpus",
    "pinjam buku": "pinjam",
    "baca buku": "pustaka",
    "kerja praktik": "pkl",
    "praktik kerja": "pkl",
    "pendaftaran": "daftar",
    "transkrip nilai": "transkrip",
}

def chatbot_kunci_mandiri(pertanyaan):
    # a. Normalisasi teks: bersihkan tanda baca dan jadikan lowercase
    kalimat = pertanyaan.lower()
    for tanda in ["?", "!", ".", ",", ":", ";", "(", ")", "-"]:
        kalimat = kalimat.replace(tanda, " ")
        
    # b. Substitusi frasa sinonim (urutkan dari frasa terpanjang)
    for frasa in sorted(kamus_sinonim_mandiri.keys(), key=len, reverse=True):
        if frasa in kalimat:
            kalimat = kalimat.replace(frasa, kamus_sinonim_mandiri[frasa])
            
    kata = kalimat.split()
    
    skor_terbaik = 0
    jawaban_terbaik = "Maaf, pertanyaan Anda belum tercantum dalam FAQ resmi. Silakan hubungi BAAK."
    kata_cocok = []

    # c. Pencocokan kata kunci dengan daftar FAQ
    for item in faq_mandiri:
        cocok = [k for k in item["kunci"] if k in kata]
        if len(cocok) > skor_terbaik:
            skor_terbaik = len(cocok)
            jawaban_terbaik = item["jawab"]
            kata_cocok = cocok

    return jawaban_terbaik, skor_terbaik, kata_cocok

# 3. Pengujian: Menguji Pertanyaan yang Sebelumnya Gagal + 3 FAQ Baru
skenario_latihan_1 = [
    "Bagaimana cara mengambil rencana studi?",               
    "Berapa batas waktu peminjaman buku di perpustakaan?",   
    "Kapan jadwal pendaftaran kerja praktik dibuka?",        
    "Bagaimana prosedur legalisir ijazah dan transkrip?",    
    "Di mana lokasi warung makan terenak dekat kampus?",    
]

print("=== PENGUJIAN CHATBOT KATA KUNCI LATIHAN MANDIRI 1 ===\n")
for q in skenario_latihan_1:
    jawab, skor, kata = chatbot_kunci_mandiri(q)
    print(f"Mahasiswa : {q}")
    print(f"Chatbot   : {jawab}")
    print(f"Status    : Skor={skor} | Kata Kunci Cocok: {kata}\n")
```
