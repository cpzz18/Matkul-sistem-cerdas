## Module 4 : Metode Pencarian: BFS, DFS, dan A*

### Kegiatan 1 — Membangun Graf dengan Dictionary

```python
peta = {
    "Madiun": ["Ngawi", "Ponorogo", "Nganjuk"],
    "Ngawi": ["Madiun"],
    "Ponorogo": ["Madiun"],
    "Nganjuk": ["Madiun", "Kediri", "Jombang"],
    "Kediri": ["Nganjuk", "Jombang", "Malang"],
    "Jombang": ["Nganjuk", "Kediri", "Mojokerto"],
    "Mojokerto": ["Jombang", "Surabaya", "Malang"],
    "Malang": ["Kediri", "Mojokerto", "Surabaya"],
    "Surabaya": ["Mojokerto", "Malang"],
}

print("Tetangga Madiun:", peta["Madiun"])
```

#### Pengembangan 1

```python
from collections import deque

def cari_rute(asal, tujuan):
    antrean = deque([[asal]])
    dikunjungi = {asal}

    while antrean:
        jalur = antrean.popleft()
        kota_sekarang = jalur[-1]

        if kota_sekarang == tujuan:
            return jalur

        for tetangga in peta.get(kota_sekarang, []):
            if tetangga not in dikunjungi:
                dikunjungi.add(tetangga)
                antrean.append(jalur + [tetangga])
    return None

asal = "Madiun"
tujuan = "Surabaya"
rute = cari_rute(asal, tujuan)

print(f"Rute terpendek {asal} ke {tujuan}:")
print(" -> ".join(rute))
print(f"Jumlah langkah: {len(rute) - 1}")
```

### Kegiatan 2 — BFS: Rute dengan Langkah Tersedikit

```python
def bfs(graf, awal, tujuan):
    antrian = [[awal]]
    while antrian:
        jalur = antrian.pop(0)
        kota = jalur[-1]
        if kota  == tujuan:
            return jalur
        for tetangga in graf[kota]:
            if tetangga not in jalur:
                antrian.append(jalur + [tetangga])
    return None
rute = bfs(peta, "Madiun", "Surabaya")
print(" -> ".join(rute))
```

#### Pengembangan 2

```python
def bfs_semua_rute(graf, awal, tujuan):
    antrian = [[awal]]
    semua_rute = []

    while antrian:
        jalur = antrian.pop(0)
        kota = jalur[-1]

        if kota == tujuan:
            semua_rute.append(jalur)
            continue

        for tetangga in graf[kota]:
            if tetangga not in jalur:
                antrian.append(jalur + [tetangga])

    return semua_rute

hasil = bfs_semua_rute(peta, "Madiun", "Surabaya")

print(f"Ditemukan {len(hasil)} alternatif rute:")
for i, rute in enumerate(hasil, 1):
    langkah = len(rute) - 1
    print(f"Rute {i} ({langkah} langkah): {' -> '.join(rute)}")
```

### Kegiatan 3 — DFS: Ubah Satu Baris

```python
def dfs(graf, awal, tujuan):
    tumpukan = [[awal]]
    while tumpukan:
        jalur = tumpukan.pop()
        kota = jalur[-1]
        if kota == tujuan:
            return jalur
        for tetangga in graf[kota]:
            if tetangga not in jalur:
                tumpukan.append(jalur + [tetangga])
    return None
print(" -> ".join(dfs(peta, "Madiun", "Surabaya")))
```

#### Pengembangan 3

```python
def dfs_rekursif(graf, awal, tujuan, jalur=None):
    if jalur is None:
        jalur = [awal]

    if awal == tujuan:
        return jalur

    for tetangga in graf.get(awal, []):
        if tetangga not in jalur:
            hasil = dfs_rekursif(graf, tetangga, tujuan, jalur + [tetangga])
            if hasil:
                return hasil
    return None

rute = dfs_rekursif(peta, "Madiun", "Surabaya")
print("Rute DFS Rekursif:")
print(" -> ".join(rute))
```

### Kegiatan 4 — A*: Mencari Rute Terpendek dalam Kilometer

```python
jarak = {
    "Madiun": {"Ngawi": 32, "Ponorogo": 29, "Nganjuk": 47},
    "Ngawi": {"Madiun": 32},
    "Ponorogo": {"Madiun": 29},
    "Nganjuk": {"Madiun": 47, "Kediri": 28, "Jombang": 40},
    "Kediri": {"Nganjuk": 28, "Jombang": 39, "Malang": 79},
    "Jombang": {"Nganjuk": 40, "Kediri": 39, "Mojokerto": 24},
    "Mojokerto": {"Jombang": 24, "Surabaya": 45, "Malang": 61},
    "Malang": {"Kediri": 79, "Mojokerto": 61, "Surabaya": 94},
    "Surabaya": {"Mojokerto": 45, "Malang": 94},
}

h = {
    "Madiun": 140,
    "Ngawi": 155,
    "Ponorogo": 160,
    "Nganjuk": 100,
    "Kediri": 95,
    "Jombang": 60,
    "Mojokerto": 40,
    "Malang": 65,
    "Surabaya": 0,
}
```

```python
def a_star(jarak, h, awal, tujuan):
    terbuka = [(h[awal], 0, [awal])]

    while terbuka:
        terbuka.sort() 
        f, g, jalur = terbuka.pop(0) 
        kota = jalur[-1]

        if kota == tujuan:
            return jalur, g  

        for tetangga, km in jarak[kota].items():
            if tetangga not in jalur:
                g_baru = g + km 
                f_baru = g_baru + h[tetangga] 
                terbuka.append((f_baru, g_baru, jalur + [tetangga]))

    return None, None

rute, total = a_star(jarak, h, "Madiun", "Surabaya")

if rute:
    print(" -> ".join(rute), f"| total {total} km")
else:
    print("Rute tidak ditemukan")
```

#### Pengembangan 4

```python
def a_star_detail(jarak, h, awal, tujuan):
    terbuka = [(h[awal], 0, [awal])]
    langkah = 1

    print(f"\tProses Pencarian A* ({awal} -> {tujuan})")
    while terbuka:
        terbuka.sort()
        f, g, jalur = terbuka.pop(0)
        kota = jalur[-1]

        print(
            f"Langkah {langkah}: Evaluasi '{kota}' | g = {g} km, h = {h[kota]} km, f = {f} km"
        )
        langkah += 1

        if kota == tujuan:
            return jalur, g

        for tetangga, km in jarak[kota].items():
            if tetangga not in jalur:
                g_baru = g + km
                f_baru = g_baru + h[tetangga]
                terbuka.append((f_baru, g_baru, jalur + [tetangga]))

    return None, None

rute, total = a_star_detail(jarak, h, "Madiun", "Surabaya")

if rute:
    print("\n\tRute Optimal Ditemukan")
    print(" -> ".join(rute))
    print(f"Total Jarak: {total} km\n")

    print("Rincian Perjalanan:")
    for i in range(len(rute) - 1):
        asal_kota = rute[i]
        tujuan_kota = rute[i + 1]
        dist = jarak[asal_kota][tujuan_kota]
        print(f"  * {asal_kota} -> {tujuan_kota}: {dist} km")
else:
    print("Rute tidak ditemukan")
```

#### Latihan 1 — Perluas peta: tambahkan kota Blitar yang terhubung ke Kediri (44 km) dan Malang (78 km) pada peta, jarak, dan h (perkiraan Blitar→Surabaya ± 110 km). Jalankan ulang ketiga algoritma — pastikan tidak error.

```python
# Menambahkan kota baru Blitar dan Malang ke peta 
peta["Blitar"] = ["Kediri", "Malang"]
peta["Kediri"].append("Blitar")
peta["Malang"].append("Blitar")

# Menambahkan jarak dari Blitar ke Kediri dan Malang
jarak["Blitar"] = {"Kediri": 44, "Malang": 78}
jarak["Kediri"]["Blitar"] = 44
jarak["Malang"]["Blitar"] = 78

# Menambahkan heuristic (h) untuk Blitar
h["Blitar"] = 110

# Menampilkan hasil pencarian rute dari Madiun ke Surabaya menggunakan BFS, DFS, dan A*
print("Hasil Pencarian Madiun -> Surabaya")
rute_bfs = bfs(peta, "Madiun", "Surabaya")
rute_dfs = dfs(peta, "Madiun", "Surabaya")
rute_astar, km = a_star(jarak, h, "Madiun", "Surabaya")

# Menampilkan hasil pencarian rute dari Madiun ke Surabaya menggunakan BFS, DFS, dan A*
print("BFS :", " -> ".join(rute_bfs))
print("DFS :", " -> ".join(rute_dfs))
print(f"A*  : {' -> '.join(rute_astar)} | total {km} km")
```

#### 2. Latihan 2 — Tutup jalan: hapus ruas Mojokerto–Surabaya (misal banjir), lalu cari lagi rute Madiun → Surabaya dengan BFS dan A*. Bandingkan: apakah keduanya masih memberi rute yang sama? Berapa total km sekarang?

```python
# Menutup ruas jalan Mojokerto-Surabaya
peta["Mojokerto"].remove("Surabaya")
peta["Surabaya"].remove("Mojokerto")

# Menghapus jarak antara Mojokerto dan Surabaya
del jarak["Mojokerto"]["Surabaya"]
del jarak["Surabaya"]["Mojokerto"]

rute_bfs = bfs(peta, "Madiun", "Surabaya")
rute_astar, total_km = a_star(jarak, h, "Madiun", "Surabaya")

# Fungsi untuk menghitung total jarak rute
def hitung_km(graf_jarak, jalur):
    return sum(
        graf_jarak[jalur[i]][jalur[i + 1]] for i in range(len(jalur) - 1)
    )
# Menghitung total jarak untuk rute BFS
km_bfs = hitung_km(jarak, rute_bfs)

# Menghitung total jarak untuk rute A*
print("HASIL SETELAH RUAS MOJOKERTO-SURABAYA DITUTUP")
print(f"BFS : {' -> '.join(rute_bfs)} | total {km_bfs} km")
print(f"A*  : {' -> '.join(rute_astar)} | total {total_km} km")

print("\nApakah kedua rute sama?:", "SAMA" if rute_bfs == rute_astar else "BEDA")
print("Total KM sekarang:", km_bfs, "km (BFS) vs", total_km, "km (A*)")
```

1. **Kapan cukup pakai BFS?**  
   BFS sudah cukup dipakai kalau kita cuma peduli jumlah perpindahan kota transit paling sedikit dan tidak menghitung jarak kilometer jalannya. Waktu jalur Mojokerto–Surabaya ditutup, BFS kebetulan memilih jalur yang sama persis dengan A* karena rute lewat Malang memang rute dengan jumlah lompatan kota paling sedikit.

2. **Kapan wajib pakai A\*?**  
   A* wajib dipakai di dunia nyata seperti Google Maps, karena jarak kilometer antar-kota itu berbeda-beda. Algoritma A* cerdas karena menghitung jarak asli yang sudah dilewati ditambah perkiraan jarak sisa menuju tujuan, sehingga dijamin memilih rute dengan total kilometer paling hemat.

3. **Kenapa DFS jarang dipakai buat cari rute?**  
   DFS cara kerjanya asal masuk terus menyusuri satu cabang jalan sampai mentok baru balik arah. Akibatnya, DFS rawan memilih jalan yang memutar sangat jauh dan sama sekali tidak menjamin kita dapat rute paling cepat atau paling pendek.
