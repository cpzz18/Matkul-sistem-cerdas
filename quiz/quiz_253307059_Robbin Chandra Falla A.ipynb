### 1. Pada kasus rute Madiun→Surabaya, BFS memberi rute 248 km sedangkan A* memberi 156 km — padahal keduanya sama-sama 4 langkah. Jelaskan mengapa hal ini terjadi, dan tuliskan rumus penilaian calon pada A* beserta arti tiap komponennya.

* **Kenapa Jaraknya Beda Jauh?**
  BFS itu ibarat orang yang cuma ngitung **berapa kali pindah kota / langkah** tanpa peduli papan jarak kilometer. Bagi BFS, jalan memutar sejauh 100 km sama jalan pintas 10 km itu dianggap setara (sama-sama dihitung 1 langkah). Makanya BFS bisa milih 4 langkah yang jalurnya muter jauh (248 km). 
  
  Sebaliknya, A* jauh lebih pintar karena dia ngitung bensin/jarak fisik asli di jalan plus ngintip arah tujuan pakai bantuan kompas (heuristik). Hasilnya, A* bisa milih 4 langkah yang jalurnya benar-benar paling pendek (156 km).

* **Rumus A\* dan Artinya:**
  $$f(n) = g(n) + h(n)$$
  * $f(n)$: Total tebakan biaya keseluruhan dari kota awal sampai kota tujuan lewat titik $n$.
  * $g(n)$: Jarak asli yang sudah nyata-nyata kamu lewati dari Madiun sampai titik kamu berdiri sekarang ($n$).
  * $h(n)$ (*heuristik*): Tebakan sisa jarak lurus (seperti ditarik garis di peta) dari posisimu sekarang ($n$) langsung ke Surabaya.

### 2. Jelaskan perbedaan forward chaining dan backward chaining (arah penalaran dan situasi yang cocok untuk masing-masing), sertakan satu contoh situasi untuk tiap arah.

* **Forward Chaining (Penalaran Maju / Berangkat dari Fakta):**
  * **Arah:** Bergerak maju: dari kumpulan **fakta/gejala** yang ada $\rightarrow$ dicek pakai aturan $\rightarrow$ ketemu **kesimpulan**.
  * **Kapan Cocok Dipakai:** Saat semua bukti atau keluhan sudah terkumpul lengkap di awal dan kita mau cari tahu artinya.
  * **Contoh Gampangnya:** Memasak dari sisa isi kulkas. Kamu buka kulkas dan melihat ada telur, nasi dingin, bawang, dan kecap (fakta). Dari bahan-bahan itu, kamu menyimpulkan: "Aku bisa bikin nasi goreng."

* **Backward Chaining (Penalaran Mundur / Berangkat dari Tujuan):**
  * **Arah:** Bergerak mundur: pasang **target/dugaan dulu** di awal $\rightarrow$ mundur ke belakang untuk cek syarat pembuktiannya.
  * **Kapan Cocok Dipakai:** Saat target dugaannya sudah jelas dan kita cuma ingin membuktikan dugaan itu benar atau tidak tanpa harus memeriksa hal-hal lain yang tidak nyambung.
  * **Contoh Gampangnya:** Memastikan bisa bikin mie instan rebus. Kamu pengin makan mie rebus hangat (target). Lalu kamu mundur mengecek syaratnya ke dapur: Apakah ada sebungkus mie instan? Apakah ada air dan gas untuk merebus? Kalau semua bahan itu ada, barulah rencanamu bisa terlaksana.

### 3. Diketahui fungsi keanggotaan segitiga: nyaman (15, 25, 35) dan panas (25, 35, 45). Hitunglah derajat keanggotaan suhu 30°C pada kedua label tersebut (tuliskan caranya), lalu jelaskan arti hasilnya dalam satu-dua kalimat.

* **Himpunan Nyaman (15, 25, 35):**
  * Paling nyaman ada di suhu 25°C. Di atas itu, rasa nyamannya makin berkurang (lereng turun).
  * Posisi 30°C ada di antara rentang 25 sampai 35:
    $$\mu_{\text{nyaman}}(30) = \frac{35 - 30}{35 - 25} = \frac{5}{10} = 0,5$$

* **Himpunan Panas (25, 35, 45):**
  * Mulai terasa panas dari 25°C (skor 0) dan makin gerah menuju puncaknya di 35°C (lereng naik).
  * Posisi 30°C ada di antara rentang 25 sampai 35:
    $$\mu_{\text{panas}}(30) = \frac{30 - 25}{35 - 25} = \frac{5}{10} = 0,5$$

* **Arti Hasilnya:**
  Suhu 30°C itu pas di tengah-tengah masa peralihan. Rasanya **setengah-setengah**: 50% masih tergolong nyaman, tapi di saat bersamaan 50% sudah mulai terasa gerah/panas.

#### Soal P1 — Pandas (10 poin)

1. Muat nilai_mahasiswa.csv ke DataFrame.
2. Buat kolom nilai_akhir = 0,3 × uts + 0,7 × uas.
3. Tampilkan rata-rata nilai_akhir per kelas.
4. Tampilkan nama dan nilai_akhir 3 mahasiswa teratas.

````python
import pandas as pd
from IPython.display import display

# 1. Muat nilai_mahasiswa.csv ke DataFrame.
url = 'https://raw.githubusercontent.com/cpzz18/Matkul-sistem-cerdas/refs/heads/main/Minggu%202/nilai_mahasiswa.csv'
df = pd.read_csv(url)

# 2. Buat kolom nilai_akhir = 0,3 × uts + 0,7 × uas.
df["nilai_akhir"] = 0.3 * df["uts"] + 0.7 * df["uas"]

# 3. Tampilkan rata-rata nilai_akhir per kelas
rata_rata = df.groupby("kelas")[["nilai_akhir"]].mean().reset_index()
print("Rata rata nilai akhir")
display(rata_rata)
print()

# 4. Tampilkan nama dan nilai_akhir 3 mahasiswa teratas.
top3 = df.nlargest(3, "nilai_akhir")[["nama", "nilai_akhir"]].reset_index(drop=True)
top3.index += 1
print("3 Mahasiswa dengan Nilai Akhir Teratas")
display(top3)
````

#### Soal P2 — BFS (10 poin)

Diberikan denah kampus dalam bentuk graf (tersedia di template): Gerbang–Parkir, Gerbang–Kantin,
Parkir–Perpustakaan, Kantin–GKB, Perpustakaan–Lab AI, GKB–Lab AI. Tulis fungsi BFS (boleh
mengetik ulang dari praktikummu) lalu cari dan cetak jalur dari Gerbang ke Lab AI beserta jumlah
langkahnya.

````python
from collections import deque

# 1. Representasi peta kampus
graf = {
    'Gerbang': ['Parkir', 'Kantin'],
    'Parkir': ['Gerbang', 'Perpustaknilai_akhiraan'],
    'Kantin': ['Gerbang', 'GKB'],
    'Perpustakaan': ['Parkir', 'Lab AI'],
    'GKB': ['Kantin', 'Lab AI'],
    'Lab AI': ['Perpustakaan', 'GKB']
}

# 2. Fungsi pencarian rute terpendek dengan BFS
def cari_rute_bfs(graf, awal, akhir):
    antrean_rute = deque([[awal]])
    lokasi_terlewat = {awal}

    while antrean_rute:
        rute_sekarang = antrean_rute.popleft()
        posisi_terakhir = rute_sekarang[-1]

        # Cek apakah sudah sampai tujuan
        if posisi_terakhir == akhir:
            return rute_sekarang

        # Cek cabang lokasi yang terhubung
        for gedung_sebelah in graf.get(posisi_terakhir, []):
            if gedung_sebelah not in lokasi_terlewat:
                lokasi_terlewat.add(gedung_sebelah)
                rute_baru = rute_sekarang + [gedung_sebelah]
                antrean_rute.append(rute_baru)
    return None

# 3. Jalankan pencarian
awal = 'Gerbang'
akhir = 'Lab AI'
rute_terpilih = cari_rute_bfs(graf, awal, akhir)

# 4. Tampilkan hasil
if rute_terpilih:
    total_langkah = len(rute_terpilih) - 1
    print("=== Jalur Tercepat (BFS) ===")
    print("Jalur Tempuh  :", " -> ".join(rute_terpilih))
    print(f"Total Langkah : {total_langkah} langkah")
else:
    print("Rute tidak ditemukan.")
````

sel markdown.

````python
def segitiga(x, a, b, c):
    if x <= a or x >= c:
        return 0.0
    elif a < x <= b:
        return (x - a) / (b - a)
    elif b < x < c:
        return (c - x) / (c - b)

# Nilai input kelembapan tanah
nilai_kelembapan = 65

# Parameter himpunan fuzzy
kering = (0, 20, 50)
normal = (30, 55, 80)
basah = (60, 85, 100)

# Hitung derajat keanggotaan
m_kering = segitiga(nilai_kelembapan, *kering)
m_normal = segitiga(nilai_kelembapan, *normal)
m_basah = segitiga(nilai_kelembapan, *basah)

# Tampilkan hasil
print(f"Hasil Perhitungan untuk Kelembapan {nilai_kelembapan}%:")
print(f"- Derajat Keanggotaan Kering : {mu_kering:.2f}")
print(f"- Derajat Keanggotaan Normal : {mu_normal:.2f}")
print(f"- Derajat Keanggotaan Basah  : {mu_basah:.2f}")
````

### Interpretasi Derajat Keanggotaan Kelembapan Tanah 65%:

1. **Kering ($\mu = 0,00$):** 
   Nilai 65% berada jauh di atas batas maksimal kondisi kering ($c = 50\%$), sehingga tanah sama sekali tidak tergolong kering.

2. **Normal ($\mu = 0,60$):** 
   Nilai 65% berada pada lereng turun himpunan normal:
   $$\mu_{\text{normal}}(65) = \frac{80 - 65}{80 - 55} = \frac{15}{25} = 0,60$$
   Artinya, kondisi tanah masih dominan memiliki karakteristik kelembapan normal sebesar 60%.

3. **Basah ($\mu = 0,20$):** 
   Nilai 65% mulai memasuki lereng naik himpunan basah:
   $$\mu_{\text{basah}}(65) = \frac{65 - 60}{85 - 60} = \frac{5}{25} = 0,20$$
   Artinya, kondisi tanah sudah mulai menunjukkan sifat basah sebesar 20%.

**Kesimpulan:** Pada tingkat kelembapan 65%, tanah berada dalam masa transisi antara normal menuju basah, dengan dominasi status **Normal (0,60)** dan sedikit tergolong **Basah (0,20)**.
