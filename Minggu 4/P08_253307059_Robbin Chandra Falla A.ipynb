### 1. Menghitung Output Neuron
Rumus: $net = (w_1 \cdot x_1) + (w_2 \cdot x_2) + b = (0,4 \cdot x_1) + (0,4 \cdot x_2) - 0,6$.  
Aktivasi undak: keluaran bernilai 1 jika $net > 0$, dan bernilai 0 jika $net \le 0$.

* Masukan (0,0): $net = 0 + 0 - 0,6 = -0,6 \le 0 \rightarrow$ Keluaran = 0
* Masukan (0,1): $net = 0 + 0,4 - 0,6 = -0,2 \le 0 \rightarrow$ Keluaran = 0
* Masukan (1,0): $net = 0,4 + 0 - 0,6 = -0,2 \le 0 \rightarrow$ Keluaran = 0
* Masukan (1,1): $net = 0,4 + 0,4 - 0,6 = 0,2 > 0 \rightarrow$ Keluaran = 1

**Kesimpulan:** Neuron ini mewakili gerbang logika **AND**, karena cuma menghasilkan nilai 1 saat kedua inputnya bernilai 1.

### 2. Latihan Perceptron Gerbang AND (2 Epoch Penuh)
Kondisi awal: $w = [0, 0]$, $b = 0$, laju belajar ($\alpha$) = 0,1.  
Aturan update saat salah ($error = target - tebakan \ne 0$):  
$w_{baru} = w_{lama} + (\alpha \cdot error \cdot x)$, $b_{baru} = b_{lama} + (\alpha \cdot error)$.

**Epoch 1 (Mulai dari $w=[0, 0]$, $b=0$):**
* Contoh (0,0), target 0: $net = 0 \rightarrow$ tebak 0 (benar) $\rightarrow w=[0, 0]$, $b=0$
* Contoh (0,1), target 0: $net = 0 \rightarrow$ tebak 0 (benar) $\rightarrow w=[0, 0]$, $b=0$
* Contoh (1,0), target 0: $net = 0 \rightarrow$ tebak 0 (benar) $\rightarrow w=[0, 0]$, $b=0$
* Contoh (1,1), target 1: $net = 0 \rightarrow$ tebak 0 (**SALAH**, $error = +1$)  
  $w_1 = 0 + (0,1 \cdot 1 \cdot 1) = 0,1$; $w_2 = 0 + (0,1 \cdot 1 \cdot 1) = 0,1$; $b = 0 + (0,1 \cdot 1) = 0,1$  
  Bobot akhir: $w=[0,1; 0,1]$, $b=0,1$
* **Jumlah salah Epoch 1: 1 kali**

**Epoch 2 (Mulai dari $w=[0,1; 0,1]$, $b=0,1$):**
* Contoh (0,0), target 0: $net = 0 + 0 + 0,1 = 0,1 > 0 \rightarrow$ tebak 1 (**SALAH**, $error = -1$)  
  $w_1 = 0,1 + (0,1 \cdot -1 \cdot 0) = 0,1$; $w_2 = 0,1 + (0,1 \cdot -1 \cdot 0) = 0,1$; $b = 0,1 + (0,1 \cdot -1) = 0$  
  Bobot: $w=[0,1; 0,1]$, $b=0$
* Contoh (0,1), target 0: $net = 0 + 0,1 + 0 = 0,1 > 0 \rightarrow$ tebak 1 (**SALAH**, $error = -1$)  
  $w_1 = 0,1 + 0 = 0,1$; $w_2 = 0,1 + (0,1 \cdot -1 \cdot 1) = 0$; $b = 0 + (0,1 \cdot -1) = -0,1$  
  Bobot: $w=[0,1; 0]$, $b=-0,1$
* Contoh (1,0), target 0: $net = 0,1 + 0 - 0,1 = 0 \le 0 \rightarrow$ tebak 0 (benar)  
  Bobot: $w=[0,1; 0]$, $b=-0,1$
* Contoh (1,1), target 1: $net = 0,1 + 0 - 0,1 = 0 \le 0 \rightarrow$ tebak 0 (**SALAH**, $error = +1$)  
  $w_1 = 0,1 + (0,1 \cdot 1 \cdot 1) = 0,2$; $w_2 = 0 + (0,1 \cdot 1 \cdot 1) = 0,1$; $b = -0,1 + (0,1 \cdot 1) = 0$  
  Bobot akhir: $w=[0,2; 0,1]$, $b=0$
* **Jumlah salah Epoch 2: 3 kali**

*Perbandingan:* Jumlah kesalahan di Epoch 2 (3 kali) justru naik dibanding Epoch 1 (1 kali). Ini wajar terjadi karena garis pemisah perceptron sedang bergeser-geser meraba batas area yang tepat sebelum nantinya konvergen sempurna di epoch ke-6.

### 3. Masalah Garis pada Gerbang XOR
* Posisi keempat titik pada koordinat Cartesius:
  * Kelas 0: titik (0,0) di pojok kiri-bawah dan (1,1) di pojok kanan-atas.
  * Kelas 1: titik (0,1) di pojok kiri-atas dan (1,0) di pojok kanan-bawah.
* **Temuan:** Kedua kelas berada saling menyilang secara diagonal. Mau ditarik garis lurus horizontal, vertikal, maupun miring dari sudut mana pun, satu garis lurus tidak akan pernah bisa memisahkan kedua titik kelas 1 dari kedua titik kelas 0. Masalah ini terbukti **tidak terpisah linear** (*non-linearly separable*), sehingga perceptron tunggal pasti gagal.

### 4. Benang Merah Evaluasi Perceptron, Regresi, dan GA
Ketiganya sama-sama punya mekanisme utama untuk **belajar dari kesalahan demi menuju kondisi terbaik**:
* **Perceptron (Bab 8):** Mengukur selisih antara tebakan dan target ($error$). Kalau salah, bobot digeser sedikit demi sedikit ke arah perbaikan.
* **Kuadrat Terkecil / Regresi (Bab 7):** Menghitung jarak melesetnya garis ke titik data asli (residual/SSE), lalu mencari nilai parameter garis ($a$ dan $b$) yang membuat total kuadrat kesalahan itu sekecil mungkin.
* **Seleksi Fitness GA (Bab 6):** Menilai kualitas tiap kandidat solusi lewat skor fitness. Solusi yang jelek atau melanggar batasan dihukum (penalti), sedangkan solusi yang berkualitas tinggi dipertahankan dan dikawinsilangkan.

Ketiganya bergerak dengan prinsip yang sama: mengubah status saat ini secara bertahap berdasarkan umpan balik kualitas atau tingkat kesalahan.

### 5. Penyebab Deep Learning Meledak di Tahun 2012
Meskipun rumus matematika backpropagation sudah ditemukan sejak 1986, deep learning baru bisa meledak pada tahun 2012 (era AlexNet) karena sebelumnya **tiga syarat utama lompatan AI** belum terpenuhi bersamaan:
* **Komputasi (Hardware):** Pada 1986, prosesor komputer masih lambat. Di tahun 2012, kartu grafis (GPU) mulai dimanfaatkan untuk pemrosesan paralel jutaan operasi matriks secara serentak.
* **Data Melimpah (Big Data):** Algoritma deep learning butuh jutaan contoh agar tidak overfit. Ketersediaan internet, media sosial, dan dataset publik berskala besar (seperti ImageNet) baru matang di era 2010-an.
* **Kematangan Algoritma & Tooling:** Hadirnya penyesuaian fungsi aktivasi modern (seperti ReLU yang mencegah masalah *vanishing gradient*) membuat pelatihan jaringan saraf dalam (banyak lapisan) akhirnya berhasil dilakukan secara stabil.
