### 1. Sebuah neuron: w = [0,4; 0,4], b = −0,6, aktivasi undak. Hitung keluarannya untuk keempat masukan gerbang (0,0),(0,1),(1,0),(1,1). Gerbang logika apa yang diwakilinya?
Rumus: $net = (w_1 \cdot x_1) + (w_2 \cdot x_2) + b = (0,4 \cdot x_1) + (0,4 \cdot x_2) - 0,6$.  
Aktivasi undak: keluaran bernilai 1 jika $net > 0$, dan bernilai 0 jika $net \le 0$.

* Masukan (0,0): $net = 0 + 0 - 0,6 = -0,6 \le 0 \rightarrow$ Keluaran = 0
* Masukan (0,1): $net = 0 + 0,4 - 0,6 = -0,2 \le 0 \rightarrow$ Keluaran = 0
* Masukan (1,0): $net = 0,4 + 0 - 0,6 = -0,2 \le 0 \rightarrow$ Keluaran = 0
* Masukan (1,1): $net = 0,4 + 0,4 - 0,6 = 0,2 > 0 \rightarrow$ Keluaran = 1

**Kesimpulan:** Neuron ini mewakili gerbang logika **AND**, karena cuma menghasilkan nilai 1 saat kedua inputnya bernilai 1.

### 2. Latih perceptron gerbang AND selama DUA epoch penuh dengan tangan (w awal [0;0], b 0, laju 0,1) — tuliskan bobot setelah tiap contoh; bandingkan jumlah kesalahan epoch 1 vs epoch 2.
Kondisi awal: $w = [0, 0]$, $b = 0$, laju belajar ($\alpha$) = 0,1.  
Aturan update saat salah ($error = target - tebakan \ne 0$):  
$w_{baru} = w_{lama} + (\alpha \cdot error \cdot x)$, $b_{baru} = b_{lama} + (\alpha \cdot error)$.

**Epoch 1 (Mulai dari $w=[0, 0]$, $b=0$):**
* Contoh (0,0), target 0: $net = 0 \rightarrow$ tebak 0 (benar) $\rightarrow w=[0, 0]$, $b=0$
* Contoh (0,1), target 0: $net = 0 \rightarrow$ tebak 0 (benar) $\rightarrow w=[0, 0]$, $b=0$
* Contoh (1,0), target 0: $net = 0 \rightarrow$ tebak 0 (benar) $\rightarrow w=[0, 0]$, $b=0$
* Contoh (1,1), target 1: $net = 0 \rightarrow$ tebak 0 (**SALAH**, $error = +1$)  
  $w_1 = 0 + (0,1 \cdot 1 \cdot 1) = 0,1$; $w_2 = 0 + (0,1 \cdot 1 \cdot 1) = 0,1$; $b = 0 + (0,1 \cdot 1) = 0,1$  
  Bobot akhir: $w=[0,1; 0,1]$, $b=0,1$
* **Jumlah salah Epoch 1: 1 kali**

**Epoch 2 (Mulai dari $w=[0,1; 0,1]$, $b=0,1$):**
* Contoh (0,0), target 0: $net = 0 + 0 + 0,1 = 0,1 > 0 \rightarrow$ tebak 1 (**SALAH**, $error = -1$)  
  $w_1 = 0,1 + (0,1 \cdot -1 \cdot 0) = 0,1$; $w_2 = 0,1 + (0,1 \cdot -1 \cdot 0) = 0,1$; $b = 0,1 + (0,1 \cdot -1) = 0$  
  Bobot: $w=[0,1; 0,1]$, $b=0$
* Contoh (0,1), target 0: $net = 0 + 0,1 + 0 = 0,1 > 0 \rightarrow$ tebak 1 (**SALAH**, $error = -1$)  
  $w_1 = 0,1 + 0 = 0,1$; $w_2 = 0,1 + (0,1 \cdot -1 \cdot 1) = 0$; $b = 0 + (0,1 \cdot -1) = -0,1$  
  Bobot: $w=[0,1; 0]$, $b=-0,1$
* Contoh (1,0), target 0: $net = 0,1 + 0 - 0,1 = 0 \le 0 \rightarrow$ tebak 0 (benar)  
  Bobot: $w=[0,1; 0]$, $b=-0,1$
* Contoh (1,1), target 1: $net = 0,1 + 0 - 0,1 = 0 \le 0 \rightarrow$ tebak 0 (**SALAH**, $error = +1$)  
  $w_1 = 0,1 + (0,1 \cdot 1 \cdot 1) = 0,2$; $w_2 = 0 + (0,1 \cdot 1 \cdot 1) = 0,1$; $b = -0,1 + (0,1 \cdot 1) = 0$  
  Bobot akhir: $w=[0,2; 0,1]$, $b=0$
* **Jumlah salah Epoch 2: 3 kali**

*Perbandingan:* Jumlah kesalahan di Epoch 2 (3 kali) justru naik dibanding Epoch 1 (1 kali). Ini wajar terjadi karena garis pemisah perceptron sedang bergeser-geser meraba batas area yang tepat sebelum nantinya konvergen sempurna di epoch ke-6.

### 3. Gambarkan keempat titik XOR pada bidang koordinat, lalu coba (dan gagal) menarik satu garis 12lurus yang memisahkan kelas 0 dan 1; jelaskan temuanmu.
* Posisi keempat titik pada koordinat Cartesius:
  * Kelas 0: titik (0,0) di pojok kiri-bawah dan (1,1) di pojok kanan-atas.
  * Kelas 1: titik (0,1) di pojok kiri-atas dan (1,0) di pojok kanan-bawah.
* **Temuan:** Kedua kelas berada saling menyilang secara diagonal. Mau ditarik garis lurus horizontal, vertikal, maupun miring dari sudut mana pun, satu garis lurus tidak akan pernah bisa memisahkan kedua titik kelas 1 dari kedua titik kelas 0. Masalah ini terbukti **tidak terpisah linear** (*non-linearly separable*), sehingga perceptron tunggal pasti gagal.

### 4. Jelaskan hubungan berikut dengan kalimatmu sendiri: aturan belajar perceptron (Bab 8) ↔ kuadrat terkecil bertahap (Bab 7) ↔ seleksi fitness (Bab 6). Apa benang merah ‘memperbaiki dari kesalahan/kualitas’ pada ketiganya?
Ketiganya sama-sama punya mekanisme utama untuk **belajar dari kesalahan demi menuju kondisi terbaik**:
* **Perceptron (Bab 8):** Mengukur selisih antara tebakan dan target ($error$). Kalau salah, bobot digeser sedikit demi sedikit ke arah perbaikan.
* **Kuadrat Terkecil / Regresi (Bab 7):** Menghitung jarak melesetnya garis ke titik data asli (residual/SSE), lalu mencari nilai parameter garis ($a$ dan $b$) yang membuat total kuadrat kesalahan itu sekecil mungkin.
* **Seleksi Fitness GA (Bab 6):** Menilai kualitas tiap kandidat solusi lewat skor fitness. Solusi yang jelek atau melanggar batasan dihukum (penalti), sedangkan solusi yang berkualitas tinggi dipertahankan dan dikawinsilangkan.

Ketiganya bergerak dengan prinsip yang sama: mengubah status saat ini secara bertahap berdasarkan umpan balik kualitas atau tingkat kesalahan.

### 5. Mengapa deep learning meledak pada 2012 padahal backpropagation ada sejak 1986? Kaitkan dengan tiga syarat lompatan AI dari Bab 1.
Meskipun rumus matematika backpropagation sudah ditemukan sejak 1986, deep learning baru bisa meledak pada tahun 2012 (era AlexNet) karena sebelumnya **tiga syarat utama lompatan AI** belum terpenuhi bersamaan:
* **Komputasi (Hardware):** Pada 1986, prosesor komputer masih lambat. Di tahun 2012, kartu grafis (GPU) mulai dimanfaatkan untuk pemrosesan paralel jutaan operasi matriks secara serentak.
* **Data Melimpah (Big Data):** Algoritma deep learning butuh jutaan contoh agar tidak overfit. Ketersediaan internet, media sosial, dan dataset publik berskala besar (seperti ImageNet) baru matang di era 2010-an.
* **Kematangan Algoritma & Tooling:** Hadirnya penyesuaian fungsi aktivasi modern (seperti ReLU yang mencegah masalah *vanishing gradient*) membuat pelatihan jaringan saraf dalam (banyak lapisan) akhirnya berhasil dilakukan secara stabil.

```python
import io
import os
from dotenv import load_dotenv
from google import genai
from IPython.display import Markdown, display
import matplotlib.pyplot as plt
from PIL import Image

# 1. Muat API key dari file .env
load_dotenv()

# 2. Inisialisasi client (otomatis mendeteksi GEMINI_API_KEY dari .env)
client = genai.Client()


def penjelasan(fig=None, instruksi_tambahan=""):
    """Mengambil grafik aktif, mengirim ke Gemini, dan menampilkan penjelasan Markdown."""
    if fig is None:
        fig = plt.gcf()

    # Simpan grafik ke RAM
    buf = io.BytesIO()
    fig.savefig(buf, format="png", bbox_inches="tight", dpi=150)
    buf.seek(0)
    img = Image.open(buf)

    # Prompt instruksi
    prompt = f"""
    Jelaskan grafik ini secara sederhana, gunakan bahasa yang mudah dimengerti, 
    hindari istilah teknis yang rumit, dan sertakan contoh atau analogi jika memungkinkan. 
    Gunakan format penjelasan yang menarik dan mudah dibaca.
    {instruksi_tambahan}
    """

    # Model resmi Google GenAI
    daftar_model = ["gemini-3.8-flash", "gemini-3.5-flash", "gemini-3.5-flash-lite"]
    response_text = None

    for model_name in daftar_model:
        try:
            response = client.models.generate_content(
                model=model_name, contents=[img, prompt]
            )
            response_text = response.text
            break
        except Exception as e:
            print(f"Percobaan dengan {model_name} gagal: {e}")
            continue
        
    if response_text:
        display(Markdown(response_text))
    else:
        print("Gagal mendapatkan penjelasan dari AI. Silakan coba lagi.")
```

```python
import pandas as pd
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt

# 1. Muat dataset
url = "https://raw.githubusercontent.com/cpzz18/Matkul-sistem-cerdas/refs/heads/main/Minggu%204/kos_madiun.csv"
dk = pd.read_csv(url)

# 2. Rekayasa fitur label target
dk["kelas_harga"] = (dk["harga"] >= 1000).map({True: "mahal", False: "murah"})

# 3. Fitur dan target
X = dk[["jarak_km", "rating"]]
y = dk["kelas_harga"]

# 4. Pembagian data dengan stratifikasi agar distribusi kelas seimbang
X_latih, X_uji, y_latih, y_uji = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# 5. Model dengan batasan kedalaman (pruning) untuk mencegah overfitting
model_pohon = DecisionTreeClassifier(max_depth=3, random_state=42)
model_pohon.fit(X_latih, y_latih)

# 6. Evaluasi Akurasi
print(f"Skor LATIH: {model_pohon.score(X_latih, y_latih):.2f}")
print(f"Skor UJI  : {model_pohon.score(X_uji, y_uji):.2f}\n")

# 7. Evaluasi Detail (Precision, Recall, F1-Score)
y_pred = model_pohon.predict(X_uji)
print("=== Laporan Klasifikasi ===")
print(classification_report(y_uji, y_pred))

# 8. Uji coba prediksi data kos baru
# Contoh: Kos A (jarak 1.2 km, rating 4.5) & Kos B (jarak 6.0 km, rating 3.8)
data_baru = pd.DataFrame({
    "jarak_km": [1.2, 6.0],
    "rating": [4.5, 3.8]
})
prediksi_baru = model_pohon.predict(data_baru)

for i, pred in enumerate(prediksi_baru):
    print(f"Kos baru #{i+1} diprediksi berkategori: {pred.upper()}")
```

Pengembangan 2

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import classification_report
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB

# 1. Muat dataset
url = "https://raw.githubusercontent.com/cpzz18/Matkul-sistem-cerdas/refs/heads/main/Minggu%206/sms_spam.csv"
df = pd.read_csv(url)

# 2. Pembagian data latih dan uji (stratifikasi agar seimbang)
X_latih, X_uji, y_latih, y_uji = train_test_split(
    df["teks"], df["label"], test_size=0.25, random_state=42, stratify=df["label"]
)

# 3. Vektorisasi (huruf kecil + unigram & bigram)
vec = CountVectorizer(lowercase=True, ngram_range=(1, 2))
X_latih_angka = vec.fit_transform(X_latih)
X_uji_angka = vec.transform(X_uji)

# 4. Pelatihan model Naive Bayes
model = MultinomialNB()
model.fit(X_latih_angka, y_latih)

# 5. Evaluasi akurasi & laporan ringkas
print(f"Skor UJI: {model.score(X_uji_angka, y_uji):.2f}\n")
y_pred = model.predict(X_uji_angka)
print(classification_report(y_uji, y_pred))

# 6. Tes prediksi pesan baru
sms_baru = [
    "Selamat Anda menang undian 50 juta rupiah hubungi nomor ini",
    "Nanti sore jadi kumpul kerja kelompok di kampus?"
]
sms_angka = vec.transform(sms_baru)
prediksi = model.predict(sms_angka)

for teks, hasil in zip(sms_baru, prediksi):
    print(f"[{hasil.upper()}] -> {teks}")
```

```python
import pandas as pd
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

# 1. Muat dataset
url = "https://raw.githubusercontent.com/cpzz18/Matkul-sistem-cerdas/refs/heads/main/Minggu%206/heart_disease.csv"
df = pd.read_csv(url)
df.columns = df.columns.str.strip()  

# 2. Deteksi otomatis format kolom
if "HeartDisease" in df.columns:
    fitur = ["Age", "RestingBP", "Cholesterol", "MaxHR"]
    target = "HeartDisease"
elif "target" in df.columns:
    fitur = ["age", "trestbps", "chol", "thalach"]
    target = "target"
else:
    kolom_angka = df.select_dtypes(include=["number"]).columns.tolist()
    target = df.columns[-1]
    fitur = [k for k in kolom_angka if k != target][:4]

print(f"Fitur X  : {fitur}")
print(f"Target y : {target}\n")

# 3. Tangani nilai kosong dan pisahkan X & y
df_bersih = df[fitur + [target]].dropna()
X = df_bersih[fitur]
y = df_bersih[target]

# 4. Split data latih & uji lalu latih model
X_latih, X_uji, y_latih, y_uji = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

model = DecisionTreeClassifier(max_depth=3, random_state=42)
model.fit(X_latih, y_latih)

# 5. Evaluasi akurasi dan confusion matrix
pred_latih = model.predict(X_latih)
pred_uji = model.predict(X_uji)

skor_latih = accuracy_score(y_latih, pred_latih)
skor_uji = accuracy_score(y_uji, pred_uji)

print("--- Evaluasi Model ---")
print(f"Skor Latih : {skor_latih:.3f}")
print(f"Skor Uji   : {skor_uji:.3f}\n")

print("Confusion Matrix:")
print(confusion_matrix(y_uji, pred_uji))

# Analisis overfitting
selisih = skor_latih - skor_uji
if selisih > 0.10:
    print(
        f"\nKesimpulan: Ada indikasi overfitting (selisih {selisih:.3f} > 0.10)."
    )
else:
    print(
        f"\nKesimpulan: Model seimbang/tidak overfitting (selisih {selisih:.3f} <= 0.10)."
    )
```

1. Latihan 1 — Stabilitas evaluasi: ulang evaluasi SMS spam dengan random_state 1, 7, dan 99.
Catat ketiga akurasinya — sama atau beda? Tulis 2 kalimat: apa artinya bagi kepercayaan pada
satu angka akurasi?

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB

url = "https://raw.githubusercontent.com/cpzz18/Matkul-sistem-cerdas/refs/heads/main/Minggu%206/sms_spam.csv"
df = pd.read_csv(url)

# Uji variasi random_state lewat perulangan ringkas
for seed in [1, 7, 99]:
    X_latih, X_uji, y_latih, y_uji = train_test_split(
        df["teks"], df["label"], test_size=0.25, random_state=seed
    )

    vec = CountVectorizer()
    X_latih_vec = vec.fit_transform(X_latih)
    X_uji_vec = vec.transform(X_uji)

    model = MultinomialNB()
    model.fit(X_latih_vec, y_latih)

    skor = accuracy_score(y_uji, model.predict(X_uji_vec))
    print(f"random_state {seed:2d} -> Akurasi: {skor * 100:.2f}%")
```

Ketiga nilai akurasi menghasilkan skor yang berbeda. Karena jumlah datanya sedikit (hanya sekitar 65 SMS), pergeseran beberapa pesan saja saat data diacak sudah cukup bikin persentase akurasinya naik-turun lumayan jauh.

(pakai for), simpan skor latih & uji, lalu gambar keduanya dalam satu grafik garis. Di kedalaman
berapa jarak keduanya mulai melebar?

```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

# 1. Muat dan siapkan data kos
url = 'https://raw.githubusercontent.com/cpzz18/Matkul-sistem-cerdas/refs/heads/main/Minggu%204/kos_madiun.csv'
dk = pd.read_csv(url)
dk["kelas_harga"] = (dk["harga"] >= 1000).map({True: "mahal", False: "murah"})

X = dk[["jarak_km", "rating"]]
y = dk["kelas_harga"]

# 2. Split data latih dan uji
X_latih, X_uji, y_latih, y_uji = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# 3. Looping kedalaman 1 sampai 8
depths = range(1, 9)
skor_latih, skor_uji = [], []

for d in depths:
    pohon = DecisionTreeClassifier(max_depth=d, random_state=42)
    pohon.fit(X_latih, y_latih)
    skor_latih.append(pohon.score(X_latih, y_latih))
    skor_uji.append(pohon.score(X_uji, y_uji))

# 4. Plot kurva
plt.figure(figsize=(8, 5))
plt.plot(depths, skor_latih, marker="o", color="green", label="Akurasi Data Latih")
plt.plot(depths, skor_uji, marker="s", color="red", label="Akurasi Data Uji")
plt.title("Kurva Overfitting Decision Tree pada Data Kos")
plt.xlabel("Kedalaman Pohon (max_depth)")
plt.ylabel("Akurasi")
plt.xticks(depths)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.tight_layout()
penjelasan()
plt.show()
```
