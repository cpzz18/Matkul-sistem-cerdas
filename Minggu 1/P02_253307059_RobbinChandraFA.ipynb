#### 1. Buat trace table BFS untuk graf: A–B, A–C, B–D, C–D, D–E (mulai A, tujuan E; urutan tetangga sesuai abjad). Tuliskan jalur akhirnya dan jumlah langkahnya.
| Langkah | Node Dievaluasi | Antrean (Queue) | Simpul Dikunjungi (Visited) | Jalur ke Simpul Saat Ini |
| :---: | :---: | :--- | :--- | :--- |
| **0 (Awal)** | — | `[A]` | `{A}` | `[A]` |
| **1** | A | `[B, C]` | `{A, B, C}` | `A` |
| **2** | B | `[C, D]` | `{A, B, C, D}` | `A → B` |
| **3** | C | `[D]` *(D sudah di-queue)* | `{A, B, C, D}` | `A → C` |
| **4** | D | `[E]` | `{A, B, C, D, E}` | `A → B → D` |
| **5** | E *(Goal)* | `[]` | `{A, B, C, D, E}` | `A → B → D → E` |

* **Jalur Akhir:** `A → B → D → E`
* **Jumlah Langkah:** 3 langkah (transisi sisi)

#### 2. Dengan tabel jarak dan h pada Subbab 2.5, kerjakan dua langkah pertama A* untuk rute Ponorogo → Surabaya: tuliskan isi daftar terbuka beserta nilai f-nya di tiap langkah.
Saya ekstrak dulu jarak asli (*g*) antar kotanya dari tabel langkah 1 di materi, Lalu mencari selisih antara *f* dan *h*:
* **Madiun–Ponorogo:** 29 km (hasil dari nilai f=189 dikurangi nilai h=160).
* **Madiun–Nganjuk:** 47 km (hasil dari f=147 dikurangi h=100).
* **Madiun–Ngawi:** 32 km (hasil dari f=187 dikurangi h=155).

Nah, dari data jarak asli dan tabel heuristik (*h*) itu, ini dua langkah pertama *trace table* A*-nya:

| Langkah | Diproses | g | h | f | Daftar terbuka sesudahnya (urut f terkecil) |
| :---: | :--- | :---: | :---: | :---: | :--- |
| **1** | **Ponorogo** | 0 | 160 | **160** | Mad(f=169) |
| **2** | **Madiun** | 29 | 140 | **169** | Ngj(f=176), Ngw(f=216) |

**Penjelasan:**
* **Langkah 1:** Dari Ponorogo, rutenya cuma mengarah ke Madiun. Jadi Madiun masuk daftar antrean (daftar terbuka) dengan jarak asli *g* = 29, dan *h* = 140. Total *f* = 169.
* **Langkah 2:** Lalu proses Madiun karena dia satu-satunya node yang ada di daftar. Dari Madiun, jalurnya bercabang ke Nganjuk dan Ngawi:
  * **Nganjuk (Ngj):** Jarak asli yang udah ditempuh (29) ditambah jarak Madiun-Nganjuk (47) = 76. Nilai *h*-nya 100, jadi nilai *f* = 176.
  * **Ngawi (Ngw):** Jarak asli yang udah ditempuh (29) ditambah jarak Madiun-Ngawi (32) = 61. Nilai *h*-nya 155, jadi nilai *f* = 216.

#### 3. BFS menemukan rute 248 km, A* menemukan 156 km — padahal keduanya sama-sama “benar”. Jelaskan dalam bahasa sendiri apa yang sebenarnya dioptimalkan masing-masing, dan kapan jawaban BFS justru yang dibutuhkan.
* **A\*:** A* mengoptimalkan total jarak asli atau biaya nyata dari rutenya. Algoritma ini memproses jarak yang udah ditempuh ditambah sama perkiraan sisa jarak ke tujuan. Makanya dia bisa nemuin rute 156 km yang bener-bener terpendek secara fisik.
* **BFS:** BFS mengoptimalkan jumlah langkah atau lompatan paling sedikit, sistem ini ngga peduli sama jarak kilometer, pokoknya jalan 100 km sama 5 km itu dihitung sama-sama 1 langkah. Jadi, BFS milih rute 248 km karena rute itu ngelewatin kota atau transit yang lebih sedikit.
* **Kapan BFS dibutuhkan:** BFS dibutuhin waktu user lebih mementingkan jumlah transit paling sedikit dibanding jarak aslinya, atau pas semua jalurnya dianggap punya nilai/bobot yang sama.

---

#### 4. Mengapa BFS boros memori sedangkan DFS hemat? Kaitkan dengan apa yang harus disimpan masing-masing di daftar tunggunya.
* **BFS:** BFS itu cara kerjanya melebar level per level. Jadi, di daftar tunggunya queue, BFS harus nyimpen semua node atau cabang yang ada di satu level secara bersamaan sebelum dia lanjut ngecek ke level di bawahnya. Makin banyak cabang graf nya, daftar tunggunya bakal makin numpuk dan bikin memori cepat bengkak.
* **DFS:** DFS cara kerjanya lurus terus ke bawah sampai mentok di satu ujung. Di daftar tunggunya stack, DFS cuma perlu nyimpen node dari satu jalur lurus yang lagi dilewatin aja saat itu. Kalau udah mentok, dia tinggal mundur backtrack dan ganti isi memorinya pakai jalur sebelahnya. Makanya DFS jauh lebih irit karena ngga perlu nginget semua cabang secara bersamaan.

#### 5. Seorang praktikan mengganti h Surabaya-nya menjadi 100 (padahal jarak sebenarnya 0 karena Surabaya adalah tujuan). Analisis: apakah h ini masih admissible? Apa akibat yang mungkin terjadi pada hasil A*?
* **Apakah h masih admissible?** Nggak admissible, syarat utama admissible heuristic itu nilai tebakan **h** nggak boleh lebih besar dari sisa jarak aslinya overestimate. Karena jarak asli dari Surabaya ke Surabaya (sebagai kota tujuan) adalah 0, sedangkan praktikan masukin angka 100 (100 > 0), maka tebakannya jadi kemahalan.
* **Akibat pada hasil A\*:** Karena **h** nya udah ngga admissible, algoritma A* bakal kehilangan jaminan buat nemuin rute yang paling optimal (terpendek). Pas sistem nemu jalan ke Surabaya, nilai **f** nya bakal kelihatan mahal karena ketambahan 100. Akibatnya, A* bakal buang-buang waktu mengevaluasi kota-kota lain di daftar antrean yang sebenernya ngga perlu dicek, atau bahkan bisa salah ngambil keputusan dengan milih rute alternatif yang jarak aslinya justru lebih jauh.

#### **Latihan 2 — Fungsi konversi_nilai(angka) yang mengembalikan huruf mutu PNM: A (>82), AB (>75), B (>67), BC (>59), C (>49), D (>39), E (sisanya) — gunakan if/elif/else**

```python
def konversi_nilai(angka):
    if angka > 82:
        return "A"
    elif angka > 75:
        return "AB"
    elif angka > 67:
        return "B"
    elif angka > 59:
        return "BC"
    elif angka > 49:
        return "C"
    elif angka > 39:
        return "D"
    else:
        return "E"

daftar_nilai = [85, 78, 70, 62, 53, 40, 35]
for n in daftar_nilai:
    print(f"Nilai: {n}, Huruf Mutu: {konversi_nilai(n)}")
```

#### **Latihan 3 — Diberikan list of dictionary berisi 5 mahasiswa: hitung rata-rata kelas, lalu cetak nama mahasiswa yang nilainya di atas rata-rata beserta huruf mutunya (gabungkan Latihan 2).**

```python
mahasiswa = [
    {"nama": "Andi", "nilai": 85},
    {"nama": "Budi", "nilai": 65},
    {"nama": "Citra", "nilai": 78},
    {"nama": "Dewi", "nilai": 55},
    {"nama": "Eko", "nilai": 72},
]

total_nilai = sum(m["nilai"] for m in mahasiswa)
rata_rata = total_nilai / len(mahasiswa)

print(f"Rata-rata Nilai Kelas: {rata_rata:.2f}\n")
print("Mahasiswa dengan nilai di atas rata-rata:")
for m in mahasiswa:
    if m["nilai"] > rata_rata:
        print(f"- {m['nama']}: {m['nilai']} (Mutu: {konversi_nilai(m['nilai'])})")
```
