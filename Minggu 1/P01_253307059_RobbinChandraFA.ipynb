#### 1. Jelaskan dengan kata-katamu sendiri tiga kata kunci definisi sistem cerdas (menalar, belajar, beradaptasi) dan beri satu contoh aplikasi untuk masing-masing.
* **Menalar:** Sistem sendiri memiliki kemampuan untuk memproses dari data yang telah dipelajari atau di train yang digunakan untuk membuat jawaban/respon yang logis dan sesuai apa yang kita perintahkan. 
  * *Contoh:* AI agent yang menalar dan menyarankan baris kode logika selanjutnya berdasarkan konteks kode yang sedang diketik oleh programmer.
* **Belajar:**  Sistem tidak beroperasi dengan aturan static atau di coding harcode sama programmer tapi dia memproses, mengumpulkan, atau menemukan data dari kita agar dapat memberikan respon yang baik.
  * *Contoh:* Algoritma rekomendasi e-commerce atau layanan streaming seperti netflix/tokped dll yang semakin akurat menyodorkan produk atau film yang pas seiring bertambahnya riwayat pencarianmu.
* **Beradaptasi:** Sistem mampu menyesuaikan tindakan atau keluarannya secara dinamis ketika berhadapan dengan user biasanya user melakukan input yang tak terduga atau bahasa yang kurang familiar dan disuruh menyingkat respon.  
  * *Contoh:* Karakter NPC dalam game cerdas yang secara dinamis mengubah strategi pergerakan dan serangannya ketika pemain tiba-tiba menggunakan taktik baru yang tidak biasa.

#### 2. Mengapa deep learning baru meledak pada 2012 padahal backpropagation sudah ditemukan 1986? Kaitkan dengan tiga syarat lompatan AI.
Karena ada lompatan besar dalam AI baru bisa terjadi ketika ada 3 syarat ini terpenuhi:
* Algoritma: Algoritma backpropagation sendiri sudah ditemukan dan ada sejak tahun 1986.
* Daya Komputasi: Lompatan baru bisa terjadi dikarenakan ketersediaan daya komputasi akhirnya dapat memadai berkat penggunaan GPU (Graphics Processing Unit)
* Data: Ledakan deep learning juga di dukung oleh ketersediaan data internet yang akhirnya dapat digunakan untuk melatih sistem.

#### 3. Susun deskripsi PEAS untuk sistem rekomendasi kantin kampus (menu apa yang ditawarkan ke tiap mahasiswa).
* **Performance:** Peningkatan jumlah klik atau pembelian menu yang direkomendasikan, tingkat kepuasan mahasiswa, dan kecepatan mahasiswa saat memilih menu yang di tawarkan.
* **Environment:** Mahasiswa, Dosen yang mau mencari makan, katalog ketersediaan stok menu dari setiap toko di kantin, dan kondisi jam istirahat atau waktu jam matkul.
* **Actuators:** Layar aplikasi smartphone yang menampilkan menu dan notifikasi promo makanan.
* **Sensors:** Riwayat pesanan/pencarian mahasiswa, profil pengguna, jam sistem, dan data stok kasir.

#### 5. “Semakin banyak aturan IF–THEN ditulis, semakin cerdas sistemnya.” Setujukah kamu? Bahas dengan merujuk pelajaran era sistem pakar dan kelahiran machine learning.
* **Tanggapan:** Tidak setuju.
* **Alasan:** 
  * Pada era Sistem Pakar (1980–1993), penulisan ribuan aturan IF-THEN secara manual justru memicu biaya perawatan yang membengkak. Sistem menjadi sangat kaku, sulit dirawat, mudah konflik antar-aturan, dan macet tidak bisa menangani saat berhadapan dengan situasi di luar kode tertulis.
  * Keterbatasan ini melahirkan era Machine Learning yang membalik paradigma tersebut. Kecerdasan sistem modern tidak diukur dari jumlah aturan yang diketik manual oleh manusia, melainkan dari kemampuan komputer untuk menemukan polanya sendiri secara mandiri dari data.
